```python
import numpy as np
import pandas as pd
from pandas import Series,DataFrame
```

```python
dframe1 = DataFrame({'key':['X','Z','Y','Z','X','X'],'data_set1':np.arange(6)})
dframe1
```

```python
dframe2 = DataFrame({'key':['Q','Y','Z'],'data_set_2':[1,2,3]})
dframe2
```

**intersection and all modes**

```python
pd.merge(dframe1,dframe2)
```

**match specific columns**
*column should be in both of dfames*

```python
pd.merge(dframe1,dframe2,on='key')
```

**which dataframe keys to use**

```python
pd.merge(dframe1,dframe2,on='key',how='left')
# left is dframe1
```

```python
pd.merge(dframe1,dframe2,on='key',how='right')
```

 **use both dataframe(union)**

```python
pd.merge(dframe1,dframe2,on='key',how='outer' )
```

```python
dframe3 = DataFrame({'key':['X','X','X','Y','Z','Z'],'data_set_3':range(6)})
dframe3
```

```python
dframe4 = DataFrame({'key':['Y','Y','X','X','Z'],'data_set_4':range(5)})
dframe4
```

```python
pd.merge(dframe3,dframe4)
# we don't have null value
```

```python
df_left = DataFrame({'key1':['SF','SF','LA'],
                     'key2':['one','two','one'],
                     'left_data':[10,20,30]})
df_left
```

```python
df_right=DataFrame({'key1':['SF','SF','LA','LA'],'key2':['one','one','one','two'],'right_data':[40,50,60,70]})
df_right
```

```python
pd.merge(df_left,df_right,on=['key1','key2'],how='outer')
```

```python
pd.merge(df_left,df_right,on='key1')
# pandas automatically seperate key2 that are same
# when merge them on one key the rest of data fram still had multiple columns
```

**make our own suffix not automatic**

```python
pd.merge(df_left,df_right,on='key1',suffixes=('_lefty','_righty'))
```

```python
# # import pandas testing utility
# import pandas._testing as tm
# tm.N = 3
# # Create a unpivoted function
# def unpivot(frame):
#     N, K = frame.shape

#     data = {'value':frame.values.ravel('F'),
#             'variable':np.asarray(frame.columns).repeat(N),
#             'data':np.tile(np.asarray(frame.index), K)}
    
#     # return the DataFrame
#     return DataFrame(data, columns=['data','variable','value'])
    
# # Set the DataFrame we'll be using
# dframe = unpivot(tm.makeTimeDataFrame())
```

```python
import numpy as np
import pandas as pd
from pandas import DataFrame

# Create a time DataFrame
dframe = pd.DataFrame(
    np.random.randn(3, 4),
    index=pd.date_range('2000-01-01', periods=3),
    columns=['A', 'B', 'C', 'D']
)

# Create a unpivoted function
def unpivot(frame):
    N, K = frame.shape
    
    data = {'value': frame.values.ravel('F'),
            # مقادیر ستونی را پشت سر هم میچینه
            'variable': np.asarray(frame.columns).repeat(N),
            # np.asarray:ستون هایا ایندکس ها را به آرایه تبدیل میکند
            'data': np.tile(np.asarray(frame.index), K)}
            # np.tile: تکرار کردن کل آرایه
    return DataFrame(data, columns=['data', 'variable', 'value'])

dframe = unpivot(dframe)
dframe
```

**pivot data**
we want variales for particular dates

```python
# rows/columns/value
dframe_piv = dframe.pivot(index='data', columns='variable',values='value')
dframe_piv
```
