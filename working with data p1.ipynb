Reading and writing csv files-->01

```python
import numpy as np
import pandas as pd 
from pandas import Series, DataFrame
```

**Read csv file and make it DataFrame**

```python
dframe = pd.read_csv('csv_file.csv')
```

```python
dframe
```

```python
# header not be the first row
dframe = pd.read_csv('csv_file.csv',header=None)
dframe
```

**other ways of read csv**

```python
dframe = pd.read_table('csv_file.csv',sep=',',header=None)
dframe
```

**indicate particular rows to read**

```python
pd.read_csv('csv_file.csv',header=None,nrows=2)
```

**export DataFrame to csv**

```python
dframe.to_csv('mytextdata_out.csv')
```

**other delimiters**

```python
# for showing output
import sys
```

```python
dframe.to_csv(sys.stdout)
```

```python
dframe.to_csv(sys.stdout,sep='_')
```

```python
dframe.to_csv(sys.stdout,sep='?')
```

**write specific columns to csv**

```python
dframe.to_csv(sys.stdout,columns=[0,1,2])
```
