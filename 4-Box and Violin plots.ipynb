**Box Plot:** Summarizes data using quartiles, median, whiskers, and outliers.

**Violin Plot:** Combines a box plot with KDE to show both summary statistics and the shape/density of the distribution.

Box Plot: Use when you want a quick summary of the data, especially median, quartiles, spread, and outliers.

Violin Plot: Use when you want to see the shape and density of the distribution, especially when looking for skewness or multiple peaks.

```python
import numpy as np
import pandas as pd
from numpy.random import randn
from pandas import Series,DataFrame

from scipy import stats

import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data1 = randn(100)
data2 = randn(100)
```

```python
sns.boxplot([data1,data2])
```

**include outliers in boxplot**

```python
sns.boxplot([data1,data2],whis=np.inf)
```

**horizontal**

```python
sns.boxplot([data1,data2],vert=False)
plt.xlim(-4,4)
```

**violin plot: kde plot + box plot**

**Normal Distribution:**
A distribution where most data points are close to the mean, and fewer data points occur as we move farther away from the mean.

For example, in `stats.norm(0, 5)`, most values are close to **0**, but larger or smaller values are still possible.

0--> mean

5-->standard deviation


```python
#Normal Dist


data1 = stats.norm(0,5).rvs(100)

# Two gamma dist. Concatenated together
data2 = np.concatenate([stats.gamma(5).rvs(50)-1,
                       -1*stats.gamma(5).rvs(50)])
#Box plot both data1 and data2
sns.boxplot([data1,data2],whis=np.inf)
```

```python
sns.violinplot([data1,data2])
```

**bandwidth**

```python
sns.violinplot(data2,bw_adjust=0.01)
```

**including individual points**

The lines are individual data points

```python
sns.violinplot(data1,inner='stick')
```
