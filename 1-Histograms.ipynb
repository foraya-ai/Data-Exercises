```python
# Standard imports
import pandas as pd
import numpy as pd
from numpy.random import randn
# stats
from scipy import stats
# Plotting
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset1 = randn(100)
```

**plot histogram**

*default:10bins*

```python
plt.hist(dataset1)
```

```python
dataset2 = randn(80)
```

```python
plt.hist(dataset2,color='indianred')
```

**both histograms on same plot**

**Note**:for diff num datapoints we normalize them with density.(80,100)

```python
plt.hist(dataset1,density=True,color='indianred',alpha=0.5,bins=20)
plt.hist(dataset2,density=True,alpha=0.5,bins=20)
```
